## Introduction to convnets

**Instantiating a small convnet**

```python
from tensorflow import keras
from tensorflow.keras import layers
inputs = keras.Input(shape=(28, 28, 1))
x = layers.Conv2D(filters=32, kernel_size=3, activation="relu")(inputs)
x = layers.MaxPooling2D(pool_size=2)(x)
x = layers.Conv2D(filters=64, kernel_size=3, activation="relu")(x)
x = layers.MaxPooling2D(pool_size=2)(x)
x = layers.Conv2D(filters=128, kernel_size=3, activation="relu")(x)
x = layers.Flatten()(x)
outputs = layers.Dense(10, activation="softmax")(x)
model = keras.Model(inputs=inputs, outputs=outputs)
```

**Displaying the model's summary**

```python
model.summary()
```

**Training the convnet on MNIST images**

```python
from tensorflow.keras.datasets import mnist

(train_images, train_labels), (test_images, test_labels) = mnist.load_data()
train_images = train_images.reshape((60000, 28, 28, 1))
train_images = train_images.astype("float32") / 255
test_images = test_images.reshape((10000, 28, 28, 1))
test_images = test_images.astype("float32") / 255
model.compile(optimizer="rmsprop",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"])
model.fit(train_images, train_labels, epochs=5, batch_size=64)
```

**Evaluating the convnet**

```python
test_loss, test_acc = model.evaluate(test_images, test_labels)
print(f"Test accuracy: {test_acc:.3f}")
```

Model #6: Pretrained VGG16 with transfer learning

**Extracting the VGG16 features and corresponding labels**

```python
# Model 6: Pretrained VGG16 with transfer learning
def create_model_6():
    # Using full training set (2000 samples) - pretrained can handle more data
    train_ds = image_dataset_from_directory(
        new_base_dir / "train",
        image_size=(180, 180),
        batch_size=batch_size)

    val_ds = image_dataset_from_directory(
        new_base_dir / "validation",
        image_size=(180, 180),
        batch_size=batch_size,
        subset="validation",
        validation_split=0.5,
        seed=42)

    test_ds = image_dataset_from_directory(
        new_base_dir / "test",
        image_size=(180, 180),
        batch_size=batch_size,
        subset="validation",
        validation_split=0.5,
        seed=42)

    # Load pretrained VGG16 without top layers
    conv_base = keras.applications.VGG16(
        weights="imagenet",
        include_top=False,
        input_shape=input_shape)

    # Freeze convolutional base
    conv_base.trainable = False

    # Create new model on top
    inputs = keras.Input(shape=input_shape)
    x = data_augmentation(inputs)
    x = keras.applications.vgg16.preprocess_input(x)  # VGG16 specific preprocessing
    x = conv_base(x)
    x = layers.Flatten()(x)
    x = layers.Dense(256, activation="relu")(x)
    x = layers.Dropout(0.5)(x)
    outputs = layers.Dense(1, activation="sigmoid")(x)
    model = keras.Model(inputs, outputs)

    model.compile(loss="binary_crossentropy",
                  optimizer=keras.optimizers.RMSprop(learning_rate=2e-5),
                  metrics=["accuracy"])

    history = model.fit(
        train_ds,
        epochs=30,
        validation_data=val_ds,
        callbacks=get_callbacks("model_6"))

    # Fine-tuning (optional second phase)
    conv_base.trainable = True
    for layer in conv_base.layers[:-4]:  # Unfreeze last few layers
        layer.trainable = False

    model.compile(loss="binary_crossentropy",
                  optimizer=keras.optimizers.RMSprop(learning_rate=1e-5),
                  metrics=["accuracy"])

    history = model.fit(
        train_ds,
        epochs=10,
        validation_data=val_ds)

    test_loss, test_acc = model.evaluate(test_ds)
    return history, test_loss, test_acc

# Run Model 6
history_6, test_loss_6, test_acc_6 = create_model_6()
```

```python
import numpy as np

def get_features_and_labels(dataset):
    all_features = []
    all_labels = []
    for images, labels in dataset:
        preprocessed_images = keras.applications.vgg16.preprocess_input(images)
        features = conv_base.predict(preprocessed_images)
        all_features.append(features)
        all_labels.append(labels)
    return np.concatenate(all_features), np.concatenate(all_labels)

train_features, train_labels =  get_features_and_labels(train_dataset)
val_features, val_labels =  get_features_and_labels(validation_dataset)
test_features, test_labels =  get_features_and_labels(test_dataset)
```
